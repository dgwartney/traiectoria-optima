```python
from pathlib import Path
from flight_planner import Dijkstra, BFS
from flight_planner.loaders import load_flight_planner
```

```python
DATA = Path('/content/traiectoria-optima/data/processed')
planner = load_flight_planner(DATA / 'airports.csv', DATA / 'routes.csv')

print(f'{len(planner.vertices)} airports, {len(planner.edges)} routes')

distance, legs = planner.find_shortest_route('SFO', 'BOS', Dijkstra())
print(f'SFO -> BOS: {distance:.1f} km in {len(legs)} leg(s)')
for leg in legs:
    print(f'  {leg.airline} {leg.origin.iata_code} -> {leg.destination.iata_code} ({leg.distance_km:.1f} km)')
```
