```python
import folium

# 1. Create base map
m = folium.Map(location=[37.7749, -122.4194], zoom_start=12)

# 2. Create feature groups
group1 = folium.FeatureGroup(name="Parks")
group2 = folium.FeatureGroup(name="Cafes")

# 3. Add elements to groups
folium.Marker([37.769, -122.446], popup="Golden Gate Park").add_to(group1)
folium.Marker([37.775, -122.418], popup="Local Cafe").add_to(group2)

# 4. Add groups to map
m.add_child(group1)
m.add_child(group2)

# 5. Add the interactive toggle control
folium.LayerControl().add_to(m)

m
```
